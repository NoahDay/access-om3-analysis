```python
def available_variables(datastore):
    """Return a pandas dataframe summarising the variables in a datastore"""
    variable_columns = [col for col in datastore.df.columns if "variable" in col]
    return (
        datastore.df[variable_columns]
        .explode(variable_columns)
        .drop_duplicates()
        .set_index("variable")
        .sort_index()
    )
```

```python
def _get_mask(ds, definition, hemisphere, threshold=None):
    # Set defaults for each definition
    defaults = {'wave_sig_ht': 0.05, 'aice': 0.8, 'fsdrad': 200}
    if threshold is None:
        try:
            threshold = defaults[definition]
        except KeyError:
            raise ValueError(f"Unknown definition: {definition}")

    # Example for aice
    aice_var = _get_var(ds, ['aice_m', 'aice', 'ICE'])
    wave_var = _get_var(ds, ['wave_sig_ht_m', 'wave_sig_ht', 'HS'])
    fsd_var = _get_var(ds, ['fsdrad_m', 'fsdrad', 'ICEF'])
    
    # Hemisphere mask
    if hemisphere == 'south':
        hemi_mask = ds['TLAT'] < 0
    elif hemisphere == 'north':
        hemi_mask = ds['TLAT'] > 0
    else:
        raise ValueError("hemisphere must be 'north' or 'south'")

    # Base ice presence mask
    base_mask = ds[aice_var] > 0.15

    # Definition-specific condition
    if definition == 'wave_sig_ht':
        cond = ds[wave_var] > threshold
    elif definition == 'aice':
        cond = ds[aice_var] < threshold
    elif definition == 'fsdrad':
        cond = ds[fsd_var] < threshold
    elif definition in ['SIE', 'SIA']:
        cond = base_mask
    else:
        raise ValueError(f"Unknown definition: {definition}")

    return base_mask & cond & hemi_mask

def _get_var(ds, names):
    """Return the first matching variable name in ds."""
    for n in names:
        if n in ds:
            return n
    raise KeyError(f"None of the variables {names} found in dataset.")

def _integrate_area(ds, mask, definition):
    if definition == "SIA":
        aice_var = _get_var(ds, ['aice_m', 'aice', 'ICE'])
        return ds['tarea']*ds[aice_var].where(mask).sum(dim=['nx', 'ny'])
    else:
        return ds['tarea'].where(mask).sum(dim=['nx', 'ny'])

def calculate_area(ds_cice, definition='wave', threshold=0.05, integration='simple', 
                        measure='median', mask=True, method='naive', hemisphere='south'):
    mask_2d = _get_mask(ds_cice, definition, hemisphere, threshold)
    
    sea_ice_extent = _integrate_area(ds_cice, mask_2d, definition) * 1e-12

    if definition in ["wave_sig_ht", "aice", "fsdrad"]:
        name = "MIZ area"
    elif definition == "SIE":
        name = "Sea ice extent"
    elif definition == "SIA":
        name = "Sea ice area"

    return xr.DataArray(
        sea_ice_extent,
        dims=sea_ice_extent.dims,
        coords={'month': ds_cice['month']},
        name=name,
        attrs={
            'units': '10^6 km^2',
            'description': f'MIZ extent ({definition})'
        }
    )
```

```python
def _integrate_width(width, ds, method):
    if method == 'simple':
        return width.sum(dim='ny', skipna=True)
    elif method == 'weighted':
        aice_var = _get_var(ds, ['aice_m', 'aice', 'ICE'])
        return (width * ds[aice_var]).sum(dim='ny', skipna=True)
    else:
        raise ValueError(f"Unknown integration: {method}")

def _reduce_width(width, method, mask):
    if mask:
        width = width.where(width != 0)
    if method == 'median':
        return width.median(dim='nx', skipna=True)
    elif method == 'mean':
        return width.mean(dim='nx', skipna=True)
    else:
        raise ValueError(f"Unknown measure: {method}")

def calculate_miz_width(ds_cice, definition='wave', threshold=0.3, integration='simple', 
                        measure='median', mask=True, hemisphere='south', freq='monthly'):
    mask_2d = _get_mask(ds_cice, definition, hemisphere, threshold)
    
    miz_width_2d = ds_cice['HTE'].where(mask_2d) / 1000 # m to km
    miz_width_1d = _integrate_width(miz_width_2d, ds_cice, integration)
    miz_width_reduced = _reduce_width(miz_width_1d, measure, mask)

    if freq == 'monthly':
        coords = {'month': ds_cice['month']}
    elif freq == 'dayofyear':
        coords = {'dayofyear': ds_cice['dayofyear']}
    elif freq == 'daily':
        coords = {'time': ds_cice['time']}

    return xr.DataArray(
        miz_width_reduced,
        dims=miz_width_reduced.dims,
        coords=coords,
        name='miz_width',
        attrs={
            'units': 'km',
            'description': f'MIZ width ({definition})'
        }
    )
```

```python
def calculate_miz_width_spatial(
    ds_cice,
    definition='wave',
    threshold=0.3,
    integration='simple',
    measure='median',
    mask=True,
    hemisphere='south',
    freq='monthly'
):
    mask_2d = _get_mask(ds_cice, definition, hemisphere, threshold)

    miz_width_2d = ds_cice['HTE'].where(mask_2d) / 1000  # m to km
    miz_width_1d = _integrate_width(miz_width_2d, ds_cice, integration)
    miz_width_reduced = _reduce_width(miz_width_1d, measure, mask)

    if freq == 'monthly':
        coords = {'month': ds_cice['month']}
    elif freq == 'dayofyear':
        coords = {'dayofyear': ds_cice['dayofyear']}
    elif freq == 'daily':
        coords = {'time': ds_cice['time']}
    else:
        coords = {}

    # Add longitude coordinate if present
    if 'TLON' in ds_cice:
        lon = ds_cice['TLON']
        for dim in miz_width_1d.dims:
            if dim in lon.dims:
                coords['lon'] = ('nx', ds_cice['TLON'].isel(ny=50).values)
                break
    elif 'lon' in ds_cice.coords:
        coords['lon'] = ds_cice['lon']

    return xr.DataArray(
        miz_width_1d,
        dims=miz_width_1d.dims,
        coords=coords,
        name='miz_width',
        attrs={
            'units': 'km',
            'description': f'MIZ width ({definition})'
        }
    )
```

```python
def fill_nan_with_interpolation(array):
    # Create a meshgrid for the coordinates
    x = np.arange(0, array.shape[1])
    y = np.arange(0, array.shape[0])
    X, Y = np.meshgrid(x, y)
    
    # Extract valid (non-NaN) points and their values
    valid_mask = ~np.isnan(array)
    coords_valid = np.vstack((X[valid_mask], Y[valid_mask])).T
    values_valid = array[valid_mask]

    # Extract coordinates of NaN values to be filled
    coords_to_fill = np.vstack((X[~valid_mask], Y[~valid_mask])).T
    
    # Fill NaN values using griddata interpolation
    array[~valid_mask] = griddata(coords_valid, values_valid, coords_to_fill, method='linear', fill_value=np.nan)
    
    return array
```

```python
def calculate_miz_width_spatial(ds_cice, definition='wave', threshold=0.3, integration='simple', 
                        measure='median', mask=True, hemisphere='south', freq='monthly'):
    mask_2d = _get_mask(ds_cice, definition, hemisphere, threshold)
    
    miz_width_2d = ds_cice['HTE'].where(mask_2d) / 1000 # m to km
    miz_width_1d = _integrate_width(miz_width_2d, ds_cice, integration)
    miz_width_reduced = _reduce_width(miz_width_1d, measure, mask)

    if freq == 'monthly':
        coords = {'month': ds_cice['month']}
    elif freq == 'dayofyear':
        coords = {'dayofyear': ds_cice['dayofyear']}
    elif freq == 'daily':
        coords = {'time': ds_cice['time']}

    return xr.DataArray(
        miz_width_1d,
        dims=miz_width_1d.dims,
        coords=coords,
        name='miz_width',
        attrs={
            'units': 'km',
            'description': f'MIZ width ({definition})'
        }
    )
```

```python
def xr_corr_flat(a, b):
    a_flat = a.values.ravel()
    b_flat = b.values.ravel()

    valid = np.isfinite(a_flat) & np.isfinite(b_flat)

    return np.corrcoef(a_flat[valid], b_flat[valid])[0, 1]

sat = altika_on_model_grid_eff.isel(dayofyear=slice(0, n_days))
mod = spatial_miz_width_wave_d_eff.isel(dayofyear=slice(0, n_days))

sat, mod = xr.align(sat, mod, join="inner")
r_spacetime = xr_corr_flat(sat, mod)

print("Spatial-temporal correlation:", r_spacetime)
```

```python
sat_time_mean = sat.mean(dim="dayofyear")
mod_time_mean = mod.mean(dim="dayofyear")

r_temporal_mean = xr_corr_flat(sat_time_mean, mod_time_mean)

print("Correlation of temporal means:", r_temporal_mean)
```

```python
sat_space_mean = sat.mean(dim="nx")
mod_space_mean = mod.mean(dim="nx")

r_spatial_mean = xr_corr_flat(sat_space_mean, mod_space_mean)

print("Correlation of spatial means:", r_spatial_mean)
```

```python
def interp_cyclic(x_model, y_model, x_obs):
    x_model = np.asarray(x_model) % 360
    y_model = np.asarray(y_model)
    x_obs = np.asarray(x_obs) % 360

    valid = np.isfinite(x_model) & np.isfinite(y_model)
    x_model = x_model[valid]
    y_model = y_model[valid]

    order = np.argsort(x_model)
    x_model = x_model[order]
    y_model = y_model[order]

    # Add cyclic wrap points
    x_wrap = np.r_[x_model[-1] - 360, x_model, x_model[0] + 360]
    y_wrap = np.r_[y_model[-1], y_model, y_model[0]]

    return np.interp(x_obs, x_wrap, y_wrap)


def calc_stats(model_at_obs, obs):
    model_at_obs = np.asarray(model_at_obs)
    obs = np.asarray(obs)

    valid = np.isfinite(model_at_obs) & np.isfinite(obs)
    model_at_obs = model_at_obs[valid]
    obs = obs[valid]

    corr = np.corrcoef(model_at_obs, obs)[0, 1]
    rmse = np.sqrt(np.mean((model_at_obs - obs) ** 2))
    bias = np.mean(model_at_obs - obs)

    return corr, rmse, bias


def add_stats_text(ax, corr, rmse, bias, units=""):
    ax.text(
        0.02, 0.95,
        f"$r$ = {corr:.2f}\nRMSE = {rmse:.2f}{units}\nBias = {bias:.2f}{units}",
        transform=ax.transAxes,
        va="top",
        ha="left",
        bbox=dict(facecolor="white", alpha=0.75, edgecolor="none")
    )
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

def monthly_relationship_panels(df, x, y, xlabel, ylabel, title):
    df_plot_all = df[[x, y, "time"]].dropna().copy()
    df_plot_all["month"] = pd.to_datetime(df_plot_all["time"]).dt.month

    months = np.arange(1, 13)
    month_names = pd.to_datetime([f"2019-{m:02d}-01" for m in months]).strftime("%b")

    fig, axes = plt.subplots(
        3, 4,
        figsize=(16, 10),
        constrained_layout=True,
        sharex=True,
        sharey=True
    )

    axes = axes.ravel()

    lims = [
        np.nanmin([df_plot_all[x].min(), df_plot_all[y].min()]),
        np.nanmax([df_plot_all[x].max(), df_plot_all[y].max()]),
    ]

    for ax, month, month_name in zip(axes, months, month_names):
        df_plot = df_plot_all[df_plot_all["month"] == month].copy()

        ax.plot(lims, lims, "k--", linewidth=1)
        ax.set_xlim(lims)
        ax.set_ylim(lims)

        if len(df_plot) == 0:
            ax.set_title(month_name)
            ax.text(
                0.5, 0.5, "No data",
                transform=ax.transAxes,
                ha="center",
                va="center"
            )
            continue

        sns.regplot(
            data=df_plot,
            x=x,
            y=y,
            ax=ax,
            scatter_kws={"s": 10, "alpha": 0.35},
            line_kws={"color": "red", "linewidth": 1.5},
        )

        r = df_plot[x].corr(df_plot[y])
        rmse = np.sqrt(np.mean((df_plot[x] - df_plot[y]) ** 2))
        bias = np.mean(df_plot[x] - df_plot[y])

        ax.text(
            0.04,
            0.96,
            f"r = {r:.2f}\nRMSE = {rmse:.1f}\nBias = {bias:.1f}\nn = {len(df_plot)}",
            transform=ax.transAxes,
            va="top",
            ha="left",
            fontsize=8,
            bbox=dict(facecolor="white", alpha=0.8, edgecolor="none"),
        )

        ax.set_title(month_name)

    fig.supxlabel(xlabel)
    fig.supylabel(ylabel)
    fig.suptitle(title)

    plt.show()
```

```python
from scipy.stats import wasserstein_distance
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# -------------------- Config --------------------

config = {
    "hs": {
        "model": "hs",
        "obs": "swhAtMyEdge",
        "model_label": "WW3 $H_s$",
        "obs_label": "AltiKa $H_s$ at MyEdge",
        "units": "m",
        "title": "$H_s$",
        "min_value": None,
    },
    "miz": {
        "model": "mizwidth_eff",
        "obs": "mizWidthAlongTrackFromMyEdge",
        "model_label": "WW3 effective MIZ width",
        "obs_label": "AltiKa MIZ width",
        "units": "km",
        "title": "Effective MIZ width",
        "min_value": 25,
    },
}

hs_cfg = config["hs"]
miz_cfg = config["miz"]


# -------------------- Helpers --------------------

def distribution_stats(x, y, bins=50):
    x = pd.Series(x).dropna()
    y = pd.Series(y).dropna()

    wd = wasserstein_distance(x, y)

    bin_edges = np.linspace(
        min(x.min(), y.min()),
        max(x.max(), y.max()),
        bins
    )

    px, _ = np.histogram(x, bins=bin_edges, density=True)
    py, _ = np.histogram(y, bins=bin_edges, density=True)
    bin_width = np.diff(bin_edges)

    overlap = np.sum(np.minimum(px, py) * bin_width)

    return wd, overlap, bin_edges


def relationship_stats(df, xcol, ycol):
    df_plot = df[[xcol, ycol]].dropna()

    r = df_plot[xcol].corr(df_plot[ycol])
    rmse = np.sqrt(np.mean((df_plot[xcol] - df_plot[ycol]) ** 2))
    bias = np.mean(df_plot[xcol] - df_plot[ycol])

    return r, rmse, bias


def format_units(units):
    return f" [{units}]" if units else ""


# -------------------- One dataframe for all plots --------------------

required_cols = [
    hs_cfg["model"],
    hs_cfg["obs"],
    miz_cfg["model"],
    miz_cfg["obs"],
]

df_plot_all = df_aligned[required_cols].dropna().copy()

if hs_cfg["min_value"] is not None:
    df_plot_all = df_plot_all[
        (df_plot_all[hs_cfg["model"]] >= hs_cfg["min_value"]) &
        (df_plot_all[hs_cfg["obs"]] >= hs_cfg["min_value"])
    ].copy()

if miz_cfg["min_value"] is not None:
    df_plot_all = df_plot_all[
        (df_plot_all[miz_cfg["model"]] >= miz_cfg["min_value"]) &
        (df_plot_all[miz_cfg["obs"]] >= miz_cfg["min_value"])
    ].copy()


# -------------------- Variables and stats --------------------

hs_x = df_plot_all[hs_cfg["model"]]
hs_y = df_plot_all[hs_cfg["obs"]]

miz_x = df_plot_all[miz_cfg["model"]]
miz_y = df_plot_all[miz_cfg["obs"]]

hs_wd, hs_overlap, hs_bins = distribution_stats(hs_x, hs_y)
miz_wd, miz_overlap, miz_bins = distribution_stats(miz_x, miz_y)

hs_r, hs_rmse, hs_bias = relationship_stats(
    df_plot_all,
    hs_cfg["model"],
    hs_cfg["obs"]
)

miz_r, miz_rmse, miz_bias = relationship_stats(
    df_plot_all,
    miz_cfg["model"],
    miz_cfg["obs"]
)

r_model = df_plot_all[hs_cfg["model"]].corr(df_plot_all[miz_cfg["model"]])
r_obs = df_plot_all[hs_cfg["obs"]].corr(df_plot_all[miz_cfg["obs"]])

def slope_coeff(df, xcol, ycol):
    df_fit = df[[xcol, ycol]].dropna()
    slope, intercept = np.polyfit(df_fit[xcol], df_fit[ycol], 1)
    return slope

slope_model = slope_coeff(
    df_plot_all,
    hs_cfg["model"],
    miz_cfg["model"]
)

slope_obs = slope_coeff(
    df_plot_all,
    hs_cfg["obs"],
    miz_cfg["obs"]
)


# -------------------- Plot --------------------

fig, axes = plt.subplots(
    3, 2,
    figsize=(12, 13),
    constrained_layout=True
)

# Top-left: HS distribution
ax = axes[0, 0]

sns.histplot(
    hs_x,
    bins=hs_bins,
    stat="density",
    color="tab:blue",
    edgecolor=None,
    linewidth=0,
    alpha=0.25,
    ax=ax
)
sns.kdeplot(hs_x, color="tab:blue", label=hs_cfg["model_label"], ax=ax)

sns.histplot(
    hs_y,
    bins=hs_bins,
    stat="density",
    color="k",
    edgecolor=None,
    linewidth=0,
    alpha=0.25,
    ax=ax
)
sns.kdeplot(hs_y, color="k", label=hs_cfg["obs_label"], ax=ax)

ax.text(
    0.97,
    0.95,
    f"Overlap = {hs_overlap:.2f}\nWasserstein = {hs_wd:.2f} {hs_cfg['units']}",
    transform=ax.transAxes,
    ha="right",
    va="top",
    bbox=dict(facecolor="white", alpha=0.8, edgecolor="none")
)

ax.set_xlabel(f"{hs_cfg['title']}{format_units(hs_cfg['units'])}")
ax.set_ylabel("Density")
ax.legend()


# Top-right: MIZ distribution
ax = axes[0, 1]

sns.histplot(
    miz_x,
    bins=miz_bins,
    stat="density",
    color="tab:blue",
    edgecolor=None,
    linewidth=0,
    alpha=0.25,
    ax=ax
)
sns.kdeplot(miz_x, color="tab:blue", label=miz_cfg["model_label"], ax=ax)

sns.histplot(
    miz_y,
    bins=miz_bins,
    stat="density",
    color="k",
    edgecolor=None,
    linewidth=0,
    alpha=0.25,
    ax=ax
)
sns.kdeplot(miz_y, color="k", label=miz_cfg["obs_label"], ax=ax)

ax.text(
    0.97,
    0.95,
    f"Overlap = {miz_overlap:.2f}\nWasserstein = {miz_wd:.2f} {miz_cfg['units']}",
    transform=ax.transAxes,
    ha="right",
    va="top",
    bbox=dict(facecolor="white", alpha=0.8, edgecolor="none")
)

ax.set_xlabel(f"{miz_cfg['title']}{format_units(miz_cfg['units'])}")
ax.set_ylabel("Density")
ax.legend()


# Middle-left: HS relationship
ax = axes[1, 0]

sns.regplot(
    data=df_plot_all,
    x=hs_cfg["model"],
    y=hs_cfg["obs"],
    ax=ax,
    scatter_kws={"s": 10, "alpha": 0.25},
    line_kws={"color": "red", "linewidth": 2}
)

lims = [
    min(df_plot_all[hs_cfg["model"]].min(), df_plot_all[hs_cfg["obs"]].min()),
    max(df_plot_all[hs_cfg["model"]].max(), df_plot_all[hs_cfg["obs"]].max()),
]
ax.plot(lims, lims, "k--", linewidth=1)
ax.set_xlim(lims)
ax.set_ylim(lims)

ax.text(
    0.04,
    0.96,
    f"r = {hs_r:.2f}\nRMSE = {hs_rmse:.2f} {hs_cfg['units']}\nBias = {hs_bias:.2f} {hs_cfg['units']}",
    transform=ax.transAxes,
    va="top",
    ha="left",
    bbox=dict(facecolor="white", alpha=0.8, edgecolor="none")
)

ax.set_xlabel(f"{hs_cfg['model_label']}{format_units(hs_cfg['units'])}")
ax.set_ylabel(f"{hs_cfg['obs_label']}{format_units(hs_cfg['units'])}")


# Middle-right: MIZ relationship
ax = axes[1, 1]

sns.regplot(
    data=df_plot_all,
    x=miz_cfg["model"],
    y=miz_cfg["obs"],
    ax=ax,
    scatter_kws={"s": 10, "alpha": 0.25},
    line_kws={"color": "red", "linewidth": 2}
)

lims = [
    min(df_plot_all[miz_cfg["model"]].min(), df_plot_all[miz_cfg["obs"]].min()),
    max(df_plot_all[miz_cfg["model"]].max(), df_plot_all[miz_cfg["obs"]].max()),
]
ax.plot(lims, lims, "k--", linewidth=1)
ax.set_xlim(lims)
ax.set_ylim(lims)

ax.text(
    0.04,
    0.96,
    f"r = {miz_r:.2f}\nRMSE = {miz_rmse:.1f} {miz_cfg['units']}\nBias = {miz_bias:.1f} {miz_cfg['units']}",
    transform=ax.transAxes,
    va="top",
    ha="left",
    bbox=dict(facecolor="white", alpha=0.8, edgecolor="none")
)

ax.set_xlabel(f"{miz_cfg['model_label']}{format_units(miz_cfg['units'])}")
ax.set_ylabel(f"{miz_cfg['obs_label']}{format_units(miz_cfg['units'])}")


# Bottom-left: model HS vs model MIZ
ax = axes[2, 0]

sns.scatterplot(
    data=df_plot_all,
    x=hs_cfg["model"],
    y=miz_cfg["model"],
    ax=ax,
    s=10,
    alpha=0.35,
    edgecolor=None
)

sns.regplot(
    data=df_plot_all,
    x=hs_cfg["model"],
    y=miz_cfg["model"],
    ax=ax,
    scatter=False,
    line_kws={"color": "red", "linewidth": 2}
)

ax.text(
    0.04,
    0.96,
    f"r = {r_model:.2f}\nslope = {slope_model:.2f}\nn = {len(df_plot_all)}",
    transform=ax.transAxes,
    va="top",
    ha="left",
    bbox=dict(facecolor="white", alpha=0.8, edgecolor="none")
)

ax.set_xlabel(f"{hs_cfg['model_label']}{format_units(hs_cfg['units'])}")
ax.set_ylabel(f"{miz_cfg['model_label']}{format_units(miz_cfg['units'])}")


# Bottom-right: observed HS vs observed MIZ
ax = axes[2, 1]

sns.scatterplot(
    data=df_plot_all,
    x=hs_cfg["obs"],
    y=miz_cfg["obs"],
    ax=ax,
    s=10,
    alpha=0.35,
    edgecolor=None
)

sns.regplot(
    data=df_plot_all,
    x=hs_cfg["obs"],
    y=miz_cfg["obs"],
    ax=ax,
    scatter=False,
    line_kws={"color": "red", "linewidth": 2}
)

ax.text(
    0.04,
    0.96,
    f"r = {r_model:.2f}\nslope = {slope_obs:.2f}\nn = {len(df_plot_all)}",
    transform=ax.transAxes,
    va="top",
    ha="left",
    bbox=dict(facecolor="white", alpha=0.8, edgecolor="none")
)


ax.set_xlabel(f"{hs_cfg['obs_label']}{format_units(hs_cfg['units'])}")
ax.set_ylabel(f"{miz_cfg['obs_label']}{format_units(miz_cfg['units'])}")
ax.set_title(f"Observations: {hs_cfg['title']} vs {miz_cfg['title']}")

plt.show()
```

```python
from scipy.stats import pearsonr
import matplotlib.pyplot as plt
import seaborn as sns

df_south = df_aligned[df_aligned["wave_southness"] >= 0.15].copy()
df_north = df_aligned[df_aligned["wave_southness"] < 0.15].copy()

groups = {
    "Southward / southerly waves": df_south,
    "Northward / northerly waves": df_north,
}

plot_specs = [
    {
        "x": "hs",
        "y": "mizwidth_eff",
        "xlabel": "Significant wave height, Hs [m]",
        "ylabel": "MIZ width eff [km]",
    },
    {
        "x": "swhAtMyEdge",
        "y": "mizWidthAlongTrackFromMyEdge",
        "xlabel": "SWH at my edge [m]",
        "ylabel": "MIZ width along track from my edge [km]",
    },
]

fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharey=False)

for row, spec in enumerate(plot_specs):
    for col, (label, data) in enumerate(groups.items()):
        ax = axes[row, col]

        d = data[[spec["x"], spec["y"]]].dropna()

        r, p = pearsonr(d[spec["x"]], d[spec["y"]])

        sns.regplot(
            data=d,
            x=spec["x"],
            y=spec["y"],
            ax=ax,
            scatter_kws={"alpha": 0.35, "s": 20},
            line_kws={"color": "k", "linewidth": 2}
        )

        ax.text(
            0.05,
            0.95,
            f"r = {r:.2f}\np = {p:.2g}\nn = {len(d)}",
            transform=ax.transAxes,
            ha="left",
            va="top",
            bbox=dict(facecolor="white", alpha=0.8, edgecolor="none")
        )

        if row == 0:
            ax.set_title(label)

        ax.set_xlabel(spec["xlabel"])
        ax.set_ylabel(spec["ylabel"] if col == 0 else "")

plt.tight_layout()
plt.show()
```
